```python
# Calculate Pearson's linear correlation matrix
correlation_matrix = df.corr(method='pearson')

print("Pearson Correlation Matrix:")
display(correlation_matrix)
```

```python
# Plot the correlation matrix as a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Pearson Correlation Matrix')
plt.show()
```

### 4. Gráficos de Dispersão (Scatter Plots)

Agora, vamos criar os gráficos de dispersão para a maior e a menor correlação com a variável de interesse (`tempo_viagem_horas`).

*   **Maior Correlação (com `tempo_viagem_horas`):** `distancia_percorrida_km` (0.97)
*   **Menor Correlação (com `tempo_viagem_horas`):** `idade_motorista_anos` (-0.16)

```python
# Variável de interesse (y)
variable_of_interest = 'tempo_viagem_horas'

# Variável com a maior correlação com a variável de interesse (distancia_percorrida_km)
highest_corr_var = 'distancia_percorrida_km'

# Variável com a menor correlação (próxima de zero) com a variável de interesse (idade_motorista_anos)
lowest_corr_var = 'idade_motorista_anos'

# Plot para a maior correlação
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1) # 1 linha, 2 colunas, primeiro gráfico
sns.scatterplot(data=df, x=highest_corr_var, y=variable_of_interest)
plt.title(f'Dispersão: {highest_corr_var} vs {variable_of_interest} (Correlação: {correlation_matrix.loc[highest_corr_var, variable_of_interest]:.2f})')
plt.xlabel(highest_corr_var)
plt.ylabel(variable_of_interest)

# Plot para a menor correlação
plt.subplot(1, 2, 2) # 1 linha, 2 colunas, segundo gráfico
sns.scatterplot(data=df, x=lowest_corr_var, y=variable_of_interest)
plt.title(f'Dispersão: {lowest_corr_var} vs {variable_of_interest} (Correlação: {correlation_matrix.loc[lowest_corr_var, variable_of_interest]:.2f})')
plt.xlabel(lowest_corr_var)
plt.ylabel(variable_of_interest)

plt.tight_layout()
plt.show()
```

### Análise e Comparação dos Gráficos de Dispersão

*   **Gráfico de Maior Correlação (`distancia_percorrida_km` vs `tempo_viagem_horas`):**
    *   **Dispersão:** Os pontos no gráfico estão fortemente agrupados em torno de uma linha reta ascendente. À medida que `distancia_percorrida_km` aumenta, `tempo_viagem_horas` também aumenta de forma consistente. Isso visualmente confirma a **forte correlação positiva** (0.97) identificada na matriz.
    *   **Relação com o Coeficiente:** Um coeficiente de Pearson próximo de +1 indica que, conforme uma variável cresce, a outra tende a crescer em uma relação linear previsível. A linha imaginária que atravessaria a maioria dos pontos teria uma inclinação positiva acentuada, com pouca dispersão dos pontos em relação a ela.

*   **Gráfico de Menor Correlação (`idade_motorista_anos` vs `tempo_viagem_horas`):**
    *   **Dispersão:** Os pontos neste gráfico estão espalhados aleatoriamente, formando uma nuvem sem um padrão direcional claro. Não há uma tendência discernível de `tempo_viagem_horas` aumentar ou diminuir consistentemente com a `idade_motorista_anos`.
    *   **Relação com o Coeficiente:** Um coeficiente de Pearson próximo de 0 (neste caso, -0.16) indica uma **correlação linear muito fraca ou inexistente**. A dispersão ampla dos pontos e a falta de uma forma linear clara visualizam que a idade do motorista não é um bom preditor linear do tempo de viagem. Embora haja um ligeiro declínio (correlação negativa), ele é muito fraco para ser considerado significativo ou previsível.
